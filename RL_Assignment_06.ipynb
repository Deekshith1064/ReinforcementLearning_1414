```python

Open In Colab

import gymnasium as gym

def random_agent(env_name, episodes=5):
    env = gym.make(env_name)

    for episode in range(1, episodes + 1):
        state, info = env.reset()
        total_reward = 0
        steps = 0
        terminated = False
        truncated = False

        while not (terminated or truncated):
            action = env.action_space.sample()
            next_state, reward, terminated, truncated, info = env.step(action)

            total_reward += reward
            steps += 1
            state = next_state

        average_reward = total_reward / steps

        print("Episode:", episode)
        print("Total Reward:", total_reward)
        print("Average Reward:", round(average_reward, 2))
        print()

    env.close()

print("CliffWalking Environment")
random_agent("CliffWalking-v1", 5)

print("Taxi Environment")
random_agent("Taxi-v4", 5)

CliffWalking Environment
Episode: 1
Total Reward: -136438
Average Reward: -9.63

Episode: 2
Total Reward: -17695
Average Reward: -10.68

Episode: 3
Total Reward: -50816
Average Reward: -9.82

Episode: 4
Total Reward: -2506
Average Reward: -7.64

Episode: 5
Total Reward: -155535
Average Reward: -10.68

Taxi Environment
Episode: 1
Total Reward: -803
Average Reward: -4.01

Episode: 2
Total Reward: -857
Average Reward: -4.29

Episode: 3
Total Reward: -821
Average Reward: -4.11

Episode: 4
Total Reward: -839
Average Reward: -4.2

Episode: 5
Total Reward: -695
Average Reward: -3.48
```
