```python
import math

machines = ['A', 'B', 'C']

rewards = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0, 'B': 0, 'C': 0}
UCB = {'A': 0, 'B': 0, 'C': 0}

N_play = 10

for play in range(1, N_play + 1):

    if play <= 3:
        machine = machines[play - 1]

    else:
        for m in machines:
            UCB[m] = Q[m] + math.sqrt(
                (2 * math.log(play)) / N[m]
            )

        machine = max(machines, key=UCB.get)

    # Repeat reward pattern when necessary
    reward = rewards[machine][N[machine] % len(rewards[machine])]

    print("\nPlay:", play)
    print("Selected Machine:", machine)
    print("Reward:", reward)

    N[machine] += 1
    R[machine] += reward

    Q[machine] = Q[machine] + (
        (reward - Q[machine]) / N[machine]
    )

    print("Q-value:", Q[machine])

print("\nFinal Results")
print("Machine\tN\tR\tQ\tUCB")

for m in machines:
    UCB[m] = Q[m] + math.sqrt(
        (2 * math.log(N_play)) / N[m]
    )

    print(
        m, "\t",
        N[m], "\t",
        R[m], "\t",
        round(Q[m], 2), "\t",
        round(UCB[m], 2)
    )

print("\nTotal Reward =", sum(R.values()))
```
