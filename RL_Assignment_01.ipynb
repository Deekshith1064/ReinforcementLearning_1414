```python
n = int(input("Enter number of machines: "))
machines = {}


for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards separated by space: ").split()))
    machines[name] = rewards


q_values = {}


for name in machines:
    rewards = machines[name]
    q = sum(rewards) / len(rewards)
    q_values[name] = q


    print("\nMachine", name)
    print("Rewards:", rewards)
    print("Q(" + name + ") =", round(q, 2))


# Find machine with highest Q value
selected = max(q_values, key=q_values.get)


print("\nGreedy Selected Machine:", selected)
```
